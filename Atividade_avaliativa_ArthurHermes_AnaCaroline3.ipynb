```python
#02 Escreva um programa que leia 10 nomes e 10 idades. Armazene os nomes em uma lista e as idades em outra. Em seguida, exiba o nome da pessoa mais velha.

nomes = []
# lista para guardar os nomes

idades = []
# lista para guardar as idades

for i in range(10):

    nome = input("Digite o nome: ")
    nomes.append(nome)

    idade = int(input("Digite a idade: "))
    idades.append(idade)

maior_idade = idades[0]
# começa assumindo que a primeira idade é a maior

posicao = 0
# guarda a posição da maior idade

for i in range(1, 10):

    if idades[i] > maior_idade:

        maior_idade = idades[i]
        posicao = i

print("Pessoa mais velha:", nomes[posicao])
print("Idade:", maior_idade)
```

```python
#04 Solicite ao usuário uma frase. Armazene em uma lista cada palavra que começa com uma vogal. Exiba a lista resultante.


frase = input("Digite uma frase: ")
# lê a frase

palavras = frase.split()
# separa a frase em palavras

comeca_vogal = []
# lista para guardar as palavras que começam com vogal

for palavra in palavras:

    primeira_letra = palavra[0].lower()
    # pega a primeira letra da palavra e transforma em minúscula

    if primeira_letra == "a" or primeira_letra == "e" or primeira_letra == "i" or primeira_letra == "o" or primeira_letra == "u":

        comeca_vogal.append(palavra)
        # adiciona a palavra na lista

print("Palavras que começam com vogal:")
print(comeca_vogal)
```

```python
#05 Leia 10 números inteiros e armazene-os em uma lista. Em seguida, crie uma segunda lista apenas com os números primos da primeira.


numeros = []
# guarda os 10 números digitados

primos = []
# guarda apenas os números primos

for i in range(10):

    numero = int(input("Digite um número: "))
    numeros.append(numero)

for numero in numeros:

    divisores = 0
    # conta quantos divisores o número possui

    for i in range(1, numero + 1):

        if numero % i == 0:
            divisores = divisores + 1

    if divisores == 2:
        primos.append(numero)
        # adiciona o número na lista de primos

print("Lista original:")
print(numeros)

print("Números primos:")
print(primos)
```

```python
#06 Escreva um programa que leia uma lista de 8 elementos e verifique (estrutura condicional) se ela é palíndroma (lê-se igual de trás para frente).


lista = []
# guarda os 8 elementos

for i in range(8):

    elemento = input("Digite um elemento: ")
    lista.append(elemento)

palindromo = True
# assume que a lista é palíndroma

for i in range(4):
# compara início e fim da lista

    if lista[i] != lista[7 - i]:
        palindromo = False
        # encontrou diferença

if palindromo:
    print("A lista é palíndroma.")
else:
    print("A lista não é palíndroma.")
```

```python
#11 Crie uma matriz 4×4 (lista de listas) com valores fornecidos pelo usuário.
# Exiba a soma dos elementos da diagonal secundária.

matriz = []

# Lê a matriz
for i in range(4):

    linha = []

    for j in range(4):

        numero = int(input(f'Digite o valor da posição [{i}][{j}]: '))
        linha.append(numero)

    matriz.append(linha)

soma = 0

# Soma a diagonal secundária
for i in range(4):

    soma = soma + matriz[i][3 - i]

print("Matriz:")
for linha in matriz:
    print(linha)

print("Soma da diagonal secundária:", soma)
```

```python
#12 Crie uma matriz 5x5 e calcule a média de cada linha.

matriz = []
medias = []

# Lê a matriz
for i in range(5):

    linha = []

    for j in range(5):

        numero = int(input(f'Digite o valor da posição [{i}][{j}]: '))
        linha.append(numero)

    matriz.append(linha)

# Calcula a média de cada linha
for linha in matriz:

    soma = 0

    for numero in linha:
        soma = soma + numero

    media = soma / 5
    medias.append(media)

print("Matriz:")

for linha in matriz:
    print(linha)

print("Médias das linhas:")
print(medias)
```

```python
#13 Solicite uma matriz 3x3 preenchida pelo usuário.
# Verifique se ela é uma matriz identidade.

matriz = []

# Lê a matriz
for i in range(3):

    linha = []

    for j in range(3):

        numero = int(input(f'Digite o valor da posição [{i}][{j}]: '))
        linha.append(numero)

    matriz.append(linha)

identidade = True
# assume que a matriz é identidade

# Verifica os elementos
for i in range(3):

    for j in range(3):

        if i == j:
            # elementos da diagonal principal devem ser 1

            if matriz[i][j] != 1:
                identidade = False

        else:
            # os demais elementos devem ser 0

            if matriz[i][j] != 0:
                identidade = False

# Exibe o resultado
if identidade:
    print("A matriz é identidade.")
else:
    print("A matriz não é identidade.")
```

```python
#14 Faça um programa que leia uma matriz 3x3 e verifique
# se ela é simétrica (igual à sua transposta).

matriz = []

# Lê a matriz
for i in range(3):

    linha = []

    for j in range(3):

        numero = int(input(f"Digite o valor da posição [{i}][{j}]: "))
        linha.append(numero)

    matriz.append(linha)

simetrica = True
# assume que a matriz é simétrica

# Verifica se a matriz é igual à sua transposta
for i in range(3):

    for j in range(3):

        if matriz[i][j] != matriz[j][i]:

            simetrica = False

# Exibe o resultado
if simetrica:

    print("A matriz é simétrica.")

else:

    print("A matriz não é simétrica.")
```

```python
#16 Crie uma matriz 4x4 com números fornecidos pelo usuário.
# Encontre o menor valor de cada coluna e armazene em uma lista.

matriz = []

# Lê a matriz
for i in range(4):

    linha = []

    for j in range(4):

        numero = int(input(f"Digite o valor da posição [{i}][{j}]: "))
        linha.append(numero)

    matriz.append(linha)

menores = []
# guarda o menor valor de cada coluna

# Percorre as colunas
for j in range(4):

    menor = matriz[0][j]
    # assume que o primeiro elemento da coluna é o menor

    for i in range(4):

        if matriz[i][j] < menor:

            menor = matriz[i][j]

    menores.append(menor)

print("Matriz:")

for linha in matriz:
    print(linha)

print("Menores valores de cada coluna:")
print(menores)
```

```python
#17 Dada uma matriz 3x3 com valores reais,
# calcule o determinante.

matriz = []

# Lê a matriz
for i in range(3):

    linha = []

    for j in range(3):

        numero = float(input(f"Digite o valor da posição [{i}][{j}]: "))
        linha.append(numero)

    matriz.append(linha)

a = matriz[0][0]
b = matriz[0][1]
c = matriz[0][2]

d = matriz[1][0]
e = matriz[1][1]
f = matriz[1][2]

g = matriz[2][0]
h = matriz[2][1]
i = matriz[2][2]

determinante = a * (e * i - f * h) - b * (d * i - f * g) + c * (d * h - e * g)

print("Determinante:", determinante)
```

```python
#18 Faça um programa que leia duas matrizes 3x3 e calcule
# a matriz soma e a matriz diferença.

matriz1 = []
matriz2 = []

# Lê a primeira matriz
print("Primeira matriz")

for i in range(3):

    linha = []

    for j in range(3):

        numero = int(input(f"Digite o valor da posição [{i}][{j}]: "))
        linha.append(numero)

    matriz1.append(linha)

# Lê a segunda matriz
print("Segunda matriz")

for i in range(3):

    linha = []

    for j in range(3):

        numero = int(input(f"Digite o valor da posição [{i}][{j}]: "))
        linha.append(numero)

    matriz2.append(linha)

matriz_soma = []
matriz_diferenca = []

# Calcula a soma e a diferença
for i in range(3):

    linha_soma = []
    linha_diferenca = []

    for j in range(3):

        soma = matriz1[i][j] + matriz2[i][j]
        diferenca = matriz1[i][j] - matriz2[i][j]

        linha_soma.append(soma)
        linha_diferenca.append(diferenca)

    matriz_soma.append(linha_soma)
    matriz_diferenca.append(linha_diferenca)

# Exibe os resultados
print("Matriz soma:")

for linha in matriz_soma:
    print(linha)

print("Matriz diferença:")

for linha in matriz_diferenca:
    print(linha)
```

```python
#19 Solicite dois conjuntos de 10 elementos fornecidos pelo usuário.
# Mostre a interseção, união e diferença (A - B e B - A).

conjunto_a = set()
conjunto_b = set()

# Lê os elementos do conjunto A
print("Conjunto A")

for i in range(10):

    elemento = input(f"Digite o {i+1}° elemento: ")
    conjunto_a.add(elemento)

# Lê os elementos do conjunto B
print("Conjunto B")

for i in range(10):

    elemento = input(f"Digite o {i+1}° elemento: ")
    conjunto_b.add(elemento)

# Operações com conjuntos
intersecao = conjunto_a & conjunto_b
uniao = conjunto_a | conjunto_b
diferenca_ab = conjunto_a - conjunto_b
diferenca_ba = conjunto_b - conjunto_a

# Exibe os resultados
print("Interseção:")
print(intersecao)

print("União:")
print(uniao)

print("Diferença A - B:")
print(diferenca_ab)

print("Diferença B - A:")
print(diferenca_ba)
```

```python
#23 Leia dois conjuntos de inteiros.
# Verifique se um é subconjunto do outro.

conjunto_a = set()
conjunto_b = set()

print("Conjunto A")

for i in range(5):

    numero = int(input(f"Digite o {i+1}° número: "))
    conjunto_a.add(numero)

print("Conjunto B")

for i in range(5):

    numero = int(input(f"Digite o {i+1}° número: "))
    conjunto_b.add(numero)

if conjunto_a.issubset(conjunto_b):

    print("A é subconjunto de B.")

elif conjunto_b.issubset(conjunto_a):

    print("B é subconjunto de A.")

else:

    print("Nenhum conjunto é subconjunto do outro.")
```

```python
#24 Dado um conjunto de números inteiros,
# crie um novo conjunto com apenas os quadrados perfeitos encontrados.

numeros = set()
quadrados_perfeitos = set()

# Lê 10 números
for i in range(10):

    numero = int(input(f"Digite o {i+1}° número: "))
    numeros.add(numero)

# Verifica quais são quadrados perfeitos
for numero in numeros:

    for i in range(numero + 1):

        if i * i == numero:

            quadrados_perfeitos.add(numero)

print("Conjunto original:")
print(numeros)

print("Quadrados perfeitos:")
print(quadrados_perfeitos)
```

```python
#29 Crie um dicionário onde as chaves são nomes de produtos
# e os valores são listas com preços em 3 lojas diferentes.
# Encontre o produto com maior variação de preço.

produtos = {}

# Lê os dados de 5 produtos
for i in range(5):

    nome = input(f"Digite o nome do {i+1}° produto: ")

    precos = []

    for j in range(3):

        preco = float(input(f"Digite o preço da loja {j+1}: "))
        precos.append(preco)

    produtos[nome] = precos

maior_variacao = 0
produto_maior_variacao = ""

# Calcula a variação de cada produto
for nome in produtos:

    maior_preco = produtos[nome][0]
    menor_preco = produtos[nome][0]

    for preco in produtos[nome]:

        if preco > maior_preco:
            maior_preco = preco

        if preco < menor_preco:
            menor_preco = preco

    variacao = maior_preco - menor_preco

    if variacao > maior_variacao:

        maior_variacao = variacao
        produto_maior_variacao = nome

print("Produto com maior variação de preço:")
print(produto_maior_variacao)

print("Variação:")
print(maior_variacao)
```

```python
#30 Leia uma frase e armazene em um dicionário o número de ocorrências de cada palavra
# ignorando maiúsculas e minúsculas.

frase = input("Digite uma frase: ")

ocorrencias = {}

# normaliza a frase (minúsculas)
frase = frase.lower()

palavra = ""

# percorre caractere por caractere
for caractere in frase:

    if caractere != " ":
        palavra = palavra + caractere
    else:
        if palavra != "":

            if palavra in ocorrencias:
                ocorrencias[palavra] = ocorrencias[palavra] + 1
            else:
                ocorrencias[palavra] = 1

            palavra = ""

# trata a última palavra (caso não termine com espaço)
if palavra != "":

    if palavra in ocorrencias:
        ocorrencias[palavra] = ocorrencias[palavra] + 1
    else:
        ocorrencias[palavra] = 1

print("Ocorrências de palavras:")
print(ocorrencias)
```
